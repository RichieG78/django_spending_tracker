```python

# Write a .env file so load_dotenv() can read the values
with open('.env', 'w') as f:
    f.write("DEBUG=True\n")
    f.write("SECRET_KEY=abc123\n")
    f.write("ROB_TEST=time for tea\n")

print(".env file written.")
```

```python
from dotenv import load_dotenv
import os

load_dotenv()

print( os.getenv("DEBUG") )
print( os.getenv("SECRET_KEY") )
print( os.getenv("ROB_TEST") )
```
